```python
import pandas as pd
import matplotlib as plt
import datetime
```

```python
data_json = 'eva-data.json'
df = pd.read_json(data_json)
df['duration'].describe()
```

```python
df.info()
```

```python
df = df.dropna(subset=['duration','date'], how='all')
```

```python
formatted_times =[]
for x in df['duration'].values:
    print(x)
    if x = nan:
        pass
    time_and_date = datetime.datetime.strptime(x, '%H:%M')
    formatted_times.append(datetime.timedelta(hours=time_and_date.hour, minutes=time_and_date.minute, seconds=time_and_date.second).total_seconds()/(60*60))
# df['formatted_times'] = formatted_times
```

```python
plt.plot(formatted_times)
```
